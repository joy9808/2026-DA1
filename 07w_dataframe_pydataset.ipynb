```python
import numpy as np
import pandas as pd

df = pd.DataFrame([[np.nan, 2, np.nan, 0],
                   [3, 4, np.nan, 1],
                   [np.nan, np.nan, np.nan, np.nan],
                   [np.nan, 3, np.nan, 4]],
                  columns = list("ABCD"))
df
```

```python
values = {"A": 100, "B": 200, "C": 300, "D": 400}
df.fillna(value = values)
```

```python
df.fillna(value = values, limit = 2)
```

```python
from IPython.display import display, HTML

def display_side_by_side(*args, titles=None):
    html_string = "<div style='display:flex;'>"
    if titles:
        for i, df in enumerate(args):
            title_html = f"<div style='margin-right: 20px;'><h3>{titles[i]}</h3>{df.to_html()}</div>"
            html_string += title_html
    else:
        for df in args:
            html_string += f"<div style='margin-right: 20px;'>{df.to_html()}</div>"
    html_string += "</div>"
    display(HTML(html_string))

display_side_by_side(df, df.fillna(value = values, limit = 2), titles=['Original DataFrame', 'Filled DataFrame'])
```
