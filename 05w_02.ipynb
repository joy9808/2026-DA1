```python
import numpy as np
import pandas as pd

s = pd.Series([3, 1, 1, 2, None, np.nan, 4, 6, 6, np.nan])
s
```

```python
s.count()                        # 결측값 제외 원소 수: 7
```

```python
s.value_counts()                 # 고유값 빈도 (NaN 제외)
```

```python
import numpy as np
import pandas as pd

s = pd.Series([3, 1, 1, 2, None, np.nan, 4, 6, 6, np.nan])
s
```

```python
s.value_counts(sort=False, dropna=False)
```

```python
s.value_counts(bins=4)                          # 4구간으로 빈도 집계
```

```python
s.value_counts(bins=4, sort=False)              # 정렬 없이
```

```python
s.value_counts(bins=[0.5, 3.5, 6.5], sort=False)  # 경계 직접 지정
```

```python
import pandas as pd

courses = pd.Series(["Spark", "PySpark", "Hadoop"], name='courses')
courses
```

```python
fees = pd.Series([250000, 300000, 200000], name='fees')
fees
```

```python
# 방법 1: dict로 생성
df = pd.DataFrame({courses.name: courses, fees.name: fees})
df
```

```python
# 방법 2: concat으로 열 방향 합치기
df = pd.concat([courses, fees], axis=1)
df
```

```python
# 방법 3: concat 기본 (행 방향)
df = pd.concat([courses, fees])
df
```
